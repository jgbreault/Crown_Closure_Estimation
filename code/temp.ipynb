## Goal

A current estimate of forest crown closure, or an equivalent vegetation-density indicator, across a reasonably sized study area in British Columbia. It should be large enough to demonstrate a meaningful spatial problem but small enough to run on a personal computer. A workflow that:
- Identifies and retrieves appropriate public geospatial datasets
- Cleans and standardizes those datasets
- Creates a machine-learning-ready dataset
- Trains and evaluates a baseline model and a stronger candidate model

### What the model should predict and whether you would treat it as regression or classification.
Crown closure is a continuous percentage (0–100%), so regression is the natural choice — it preserves the full gradient and makes errors interpretable in meaningful units. That said, classification is also defensible if your training labels come from VRI, which stores crown closure in coarse bins (e.g., 10–25%, 26–50%, etc.). In that case, classifying on those bins is actually more honest about what the label data actually captures. I'd default to regression but acknowledge the VRI binning issue explicitly.

### What could be used as the ground-truth or target data.
The most practical source is the BC Vegetation Resource Inventory (VRI), which contains photo-interpreted crown closuretes covering all of BC — it's publicly available and directly maps to the target variable. The critical weakness: VRI capture dates vary widely, some polygons are 15–25 years old, and photo-interpretation is coarse. If any lidar coverage exists for your study area (BCTS, NRCan acquisitions), that would be a much better ground truth and worth checking. NFI/CMI field plots exist but are far too sparse for model training.

### What you would clarify with a forestry professional before beginning.
- What does "current" mean — updating VRI estimates, or specifically targeting recently disturbed areas that VRI misses?
- Which crown closure definition/measurement method should the model approximate?
- Does the model need to work across all stand types (conifer, deciduous, mixed), or just one?
- What downstream use is this for — carbon, fire behavior, timber, habitat? That sets the accuracy threshold that matters.
- Is there lidar coverage for the study area that could serve as better ground truth?

### Your key assumptions and what could make the result unsuitable for operational use
The biggest risk is label staleness: if VRI polygons in your study area are 15+ years old, you're training a "current" model on outdated targets — the model learns historical crown closure, not current. A closely related issue is spatial leakage: a naive random train/test split puts pixels from the same VRI polygon in both sets, making accuracy metrics look great while the model has actually just memorized polygon boundaries. Spatially blocked cross-validation is essential here. Other risks: NDVI saturation above ~80% crown closure (optical sensors can't distinguish dense closed-canopy), phenological mismatch if imagery spans multiple seasons, and poo-crown relationships differacross species compositions in BC.
